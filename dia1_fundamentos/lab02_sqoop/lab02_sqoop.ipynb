```python
conn = sqlite3.connect('bigdata_course.db')
pd.read_csv('customers_synthetic.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_csv('transactions_synthetic.csv').to_sql('transactions', conn, if_exists='replace', index=False)

print('clientes:', conn.execute('SELECT COUNT(*) FROM customers').fetchone()[0])
print('transações:', conn.execute('SELECT COUNT(*) FROM transactions').fetchone()[0])
```

```python
df = pd.read_sql('SELECT * FROM customers', conn)
df.to_csv('bigdata/raw/customers/customers_from_db.csv', index=False)
df2 = pd.read_sql('SELECT * FROM transactions', conn)
df2.to_csv('bigdata/raw/transactions/transactions_from_db.csv', index=False)
print("Exportado para bigdata/raw/")
```

```python
chunks = [df2[i::4] for i in range(4)]
for i, chunk in enumerate(chunks):
    chunk.to_csv(f'bigdata/raw/transactions/part-{i}.csv', index=False)
    print(f'part-{i}: {len(chunk)} linhas')
```
